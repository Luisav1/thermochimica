```python
import numpy as np
```

```python
calcs = 22500
temperature = 2900
pressure = 1
```

```python
A = np.linspace(0,1,22500)
```

```python
B = 1-A
```

```python
# X,Y = np.meshgrid(temperature,pressure)
```

```python
# ratio_1,ratio_2 = 2.05,1
```

```python
# sum = ratio_1 + ratio_2
```

```python
array = np.zeros((calcs,4))
array[:,0] = temperature
array[:,1] = pressure
array[:,2] = A
array[:,3] = B
array
```

```python
np.savetxt("array.csv", array, delimiter=" ")
```

```python
vals = array[:,2] + array[:,3]
```

```python
array[:,2] = array[:,2] / vals
array[:,3] = array[:,3] / vals
```

```python
np.savetxt('thermo.txt',array,delimiter = ' ')
```
